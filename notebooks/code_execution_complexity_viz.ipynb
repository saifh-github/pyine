# Experiment Results Visualization

This notebook visualizes the results from the code execution complexity experiments.

```python
import matplotlib.pyplot as plt

import pyine.apps.code_execution_complexity.utils.analysis as analysis
import pyine.apps.code_execution_complexity.utils.persistence as persistence
import pyine.apps.code_execution_complexity.visualization as visualization
import pyine.utils.filesystem
```

## List Available Experiments

```python
default_output_dir = pyine.utils.filesystem.get_logs_root_path() / "code_exec_complexity_results"
experiments = persistence.list_experiments(default_output_dir)

print(f"Found {len(experiments)} experiments:\n")
for i, exp in enumerate(experiments, 1):
    print(f"{i}. {exp.name} - {exp.timestamp}")
    print(f"   Path: {exp.path}")
    print()
```

## Select Experiments to Analyze

Choose one or more experiments by adding their paths (printed in the previous cell) to the list below.

```python
# Single experiment example:
# experiment_paths = [
#    experiments[0].path  # Most recent experiment
# ]

# Multiple experiments example:
experiment_paths = [e.path for e in experiments if "gpt4o" in e.name]

# Filtered by metadata example:
# filtered = analysis.filter_experiments_by_metadata(
#     experiments,
#     predictor_model="gpt-4o",
#     reasoning_effort=None
# )
# experiment_paths = [e['path'] for e in filtered]

# === LOAD AND AGGREGATE ===
df, metadata_list = analysis.load_and_aggregate_experiments(experiment_paths)
results = df.to_dict('records')  # for compatibility with visualization functions

print(f"Loaded {len(experiment_paths)} experiment(s)")
print(f"Total test cases: {len(df)}\n")
```

## Experiment Summary

```python
# Breakdown of every experiment
if len(experiment_paths) > 1:
    targeted_experiments = [e for e in experiments if e.path in experiment_paths]
    summary_df = analysis.get_experiment_summary(targeted_experiments, metadata_list)
    print("Breakdown of each experiment:")
    print(summary_df.to_string(index=False))
    print("\n" + "=" * 80 + "\n")

# Overall aggregated statistics
analysis.print_overall_summary(df, len(experiment_paths))
```

## Visualization 1: Accuracy Comparison

```python
fig = visualization.plot_accuracy_comparison(results)
plt.show()
```

## Visualization 2: Complexity Distributions

```python
fig = visualization.plot_complexity_distributions(results)
plt.show()
```

## Visualization 3: Accuracy vs Complexity Metrics

```python
for _fig in visualization.plot_all_accuracy_vs_complexity(results):
    plt.show()
```

## Visualization 4: Correlation Analysis

```python
fig = visualization.plot_correlation_analysis(results)
plt.show()

# Print the correlation table
analysis.print_correlation_table(df)
```
